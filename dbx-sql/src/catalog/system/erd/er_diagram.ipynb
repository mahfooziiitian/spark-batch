```python
from sqlalchemy.engine import create_engine
import os
from sqlalchemy import MetaData
from graphviz import Digraph
```

```python
token = os.environ.get("M_DBX_TOKEN")
host = os.environ.get("M_DBX_HOST").split("://")[1]
http_path = os.environ.get("M_DBX_HTTP_PATH")
catalog = os.environ.get("M_DBX_CATALOG", "mgmt_stg")
schema = os.environ.get("M_DBX_SCHEMA", "poc")
```

```python
url = f"databricks://token:{token}@{host}?http_path={http_path}&catalog={catalog}&schema={schema}"
```

```python
engine = create_engine(url)
```

```python
metadata = MetaData()
metadata.reflect(bind=engine)

for table in metadata.tables:
    print(table)
```

```python
# Create diagram
dot = Digraph(comment="ER Diagram", format="png")

# Add tables and columns
for table in metadata.tables.values():
    label = f"<<TABLE BORDER='0' CELLBORDER='1' CELLSPACING='0'>"
    label += f"<TR><TD COLSPAN='1'><B>{table.name}</B></TD></TR>"
    for column in table.columns:
        label += f"<TR><TD>{column.name}</TD></TR>"
    label += "</TABLE>>"
    dot.node(table.name, label=label, shape="plain")

# Add foreign key relationships
for table in metadata.tables.values():
    for fk in table.foreign_keys:
        dot.edge(table.name, fk.column.table.name)

# Save and render
dot.render("erd_diagram", view=True)
```

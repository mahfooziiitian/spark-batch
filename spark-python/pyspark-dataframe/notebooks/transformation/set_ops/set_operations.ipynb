```python
from data_frame.transformation.set_ops.set_operator import SetOperator
from data_frame.spark_utils import get_spark
```

```python
spark = get_spark(app_name="Set Operations")
```

## 1. Union Operations

```python
# Create two DataFrames
data1 = [("A", 1), ("B", 2), ("C", 3)]
data2 = [("B", 2), ("D", 4), ("E", 5)]

df1 = spark.createDataFrame(data1, ["letter", "number"])
df2 = spark.createDataFrame(data2, ["letter", "number"])

# UNION ALL (keeps duplicates)
union_all = SetOperator.union_all(df1, df2)
print("UNION ALL (keeps duplicates):")
union_all.show()

# UNION DISTINCT (removes duplicates)
union_distinct = SetOperator.union_distinct(df1, df2)
print("UNION DISTINCT (removes duplicates):")
union_distinct.show()
```

## 2. Intersect and Except

```python
# INTERSECT (common elements)
intersect_result = SetOperator.intersect(df1, df2)
print("INTERSECT (common elements):")
intersect_result.show()

# EXCEPT (elements in df1 not in df2)
except_result = SetOperator.except_df(df1, df2)
print("EXCEPT (in df1 but not in df2):")
except_result.show()
```

## 3. Multiple DataFrame Union

```python
# Create multiple DataFrames
df3 = spark.createDataFrame([("F", 6)], ["letter", "number"])
df4 = spark.createDataFrame([("G", 7)], ["letter", "number"])

# Union all DataFrames
multi_union = SetOperator.union_multiple([df1, df2, df3, df4])
print("Union of multiple DataFrames:")
multi_union.show()
```

```python
from data_frame.columns.column_transformer import ColumnTransformer
from pyspark.sql import types as T
from data_frame.spark_utils import get_spark
```

```python
spark = get_spark(app_name="Schema Inspection")
```

```python
# Create base DataFrame
data = [("Alice", 1000), ("Bob", 2000), ("Charlie", 1500)]
df = spark.createDataFrame(data, ["name", "salary"])
```

```python
# Add columns
df = ColumnTransformer.add_columns(
    df, 
    department="Engineering",
    location="New York",
    active=True
)
```

```python
# Rename columns
df = ColumnTransformer.rename_columns(
    df,
    {"name": "employee_name", "salary": "monthly_salary"}
)

df.show()
```

```python
# Select specific columns
selected_df = ColumnTransformer.select_columns(
    df, ["employee_name", "monthly_salary"]
)
selected_df.show()
```

```python
# Drop columns
dropped_df = ColumnTransformer.select_columns(
    df, ["location", "active"], drop=True
)
dropped_df.show()
```

```python
"""
## 3. Type Casting
"""
# Cast columns
casted_df = ColumnTransformer.cast_columns(
    df,
    {"monthly_salary": T.StringType(), "active": T.StringType()}
)
casted_df.printSchema()
```

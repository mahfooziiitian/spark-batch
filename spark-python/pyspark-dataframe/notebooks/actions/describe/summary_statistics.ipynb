```python
from data_frame.exploration.data_profile import DataFrameProfiler
from data_frame.spark_utils import get_spark
```

```python
spark = get_spark(app_name="Summary statistics")
```

```python
# Create sample data
sales_data = [
    (1, "Product A", 100, 25.50, "North"),
    (2, "Product B", 150, 30.00, "South"),
    (3, "Product C", 80, 15.75, "East"),
    (4, "Product A", 200, 25.50, "West"),
    (5, "Product B", 120, 30.00, "North"),
    (None, "Product C", 90, 15.75, "South")
]
```

```python
df_sales = spark.createDataFrame(sales_data, 
                                 ["id", "product", "quantity", "price", "region"])
```

```python
# Basic statistics
stats = DataFrameProfiler.basic_stats(df_sales)
stats.show()
```

```python

"""
## 2. Detailed Column Profiling
"""
profile = DataFrameProfiler.column_profiling(df_sales)
for col_name, col_stats in profile.items():
    print(f"\n{col_name}:")
    for key, value in col_stats.items():
        print(f"  {key}: {value}")
```

```python
"""
## 3. Correlation Analysis
"""
correlation = DataFrameProfiler.correlation_analysis(df_sales, "quantity", "price")
print(f"Correlation between quantity and price: {correlation:.3f}")
```

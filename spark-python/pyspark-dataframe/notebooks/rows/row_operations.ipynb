```python
from data_frame.rows.row_processor import RowProcessor
from data_frame.spark_utils import get_spark
from pyspark.sql import functions as F
```

```python
spark = get_spark(app_name="Row Operations")
```

```python
data = [
    (1, "Product A", 100, "North"),
    (2, "Product B", 200, "South"),
    (3, "Product A", 100, "North"),  # Duplicate
    (4, "Product C", 150, "East"),
    (5, "Product B", 200, "South")   # Duplicate
]
df = spark.createDataFrame(data, ["id", "product", "price", "region"])
```

```python
# Get distinct rows
distinct_df = RowProcessor.distinct_rows(df)
print("Distinct rows:")
distinct_df.show()
```

```python
# Drop duplicates based on specific columns
deduped_df = RowProcessor.drop_duplicates(df, subset=["product", "price"])
print("Deduplicated by product and price:")
deduped_df.show()
```

```python

"""
## 2. Sampling and Limiting
"""
# Random sample
sampled_df = RowProcessor.sample_rows(df, fraction=0.5, seed=42)
print("Random sample (50%):")
sampled_df.show()

# Limit rows
limited_df = RowProcessor.limit_rows(df, 2)
print("First 2 rows:")
limited_df.show()
```

```python
"""
## 3. Row Filtering
"""
# Filter by condition
filtered_df = RowProcessor.filter_by_row_condition(
    df, (F.col("price") > 100) & (F.col("region") != "South")
)
print("Price > 100 and region not South:")
filtered_df.show()
```
